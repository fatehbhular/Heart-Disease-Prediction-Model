## 2.2 Initial Data Quality Inspection

```python
# Check for missing values in each column
missing_values = df_original.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Count fully duplicated rows
print("Duplicate rows:", df_original.duplicated().sum())
```

```python
# Check the number of unique values in each column
for column in df_original.columns:
    print(f"{column}: {df_original[column].nunique()} unique values")
```

```python
# Inspect the distributions and ranges of numerical variables
df_original.describe().T
```

```python
# Inspect the values and frequencies of all categorical columns
categorical_columns = df_original.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df_original[column].value_counts(dropna=False))
```

```python
# Investigate heart_disease because it was expected to be binary
print(df_original["heart_disease"].value_counts(dropna=False))
print("Unique values:", df_original["heart_disease"].unique())
```
